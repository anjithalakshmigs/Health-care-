```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Reload original data so we can clean it properly
df = pd.read_csv("/content/health data.csv")

# Create a separate copy for cleaning
clean_df = df.copy()

# Remove duplicate rows
print("Duplicate rows:", clean_df.duplicated().sum())
clean_df = clean_df.drop_duplicates().copy()

# Handle missing age values
clean_df["age"] = pd.to_numeric(clean_df["age"], errors="coerce")
clean_df.loc[~clean_df["age"].between(0, 120), "age"] = np.nan
clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())

# Keep missing doctor IDs missing
clean_df["doctor_id"] = pd.to_numeric(
    clean_df["doctor_id"], errors="coerce"
)

# Correct patient ID data type without inventing IDs
clean_df["patient_id"] = pd.to_numeric(
    clean_df["patient_id"], errors="coerce"
).astype("Int64")

print("Shape after cleaning:", clean_df.shape)
print("\nMissing values after cleaning:")
display(clean_df.isnull().sum().to_frame("Missing Values"))

print("\nCleaned data preview:")
display(clean_df.head())
```

```python
# Create age groups
clean_df["age_band"] = pd.cut(
    clean_df["age"],
    bins=[0, 17, 29, 44, 59, 120],
    labels=["0-17", "18-29", "30-44", "45-59", "60+"],
    include_lowest=True
)

# Count symptoms, if the symptoms column exists
if "symptoms" in clean_df.columns:
    clean_df["symptom_count"] = (
        clean_df["symptoms"]
        .fillna("")
        .apply(lambda x: len([s for s in str(x).split(",") if s.strip()]))
    )

print("Age group counts:")
display(clean_df["age_band"].value_counts().sort_index())

# Filter records for patients aged 60 or above
older_patients = clean_df[clean_df["age"] >= 60]

print("Records for patients aged 60 and above:", len(older_patients))
display(older_patients.head())

# Save the cleaned dataset
clean_df.to_csv("/content/health_data_cleaned.csv", index=False)
```

```python
# Numerical summary
display(clean_df.describe().T)

# Gender distribution
print("Gender distribution:")
display(clean_df["gender"].value_counts(dropna=False))

# Diagnosis distribution
print("Diagnosis distribution:")
display(clean_df["Diagnosis"].value_counts().head(15))

# Age distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=clean_df, x="age", bins=20, kde=True)
plt.title("Distribution of Patient Ages")
plt.xlabel("Age")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

# Gender count plot
plt.figure(figsize=(7, 5))
sns.countplot(data=clean_df, x="gender")
plt.title("Patient Records by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python
# Numerical summary
display(clean_df.describe().T)

# Gender distribution
print("Gender distribution:")
display(clean_df["gender"].value_counts(dropna=False))

# Diagnosis distribution
print("Diagnosis distribution:")
display(clean_df["Diagnosis"].value_counts().head(15))

# Age distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=clean_df, x="age", bins=20, kde=True)
plt.title("Distribution of Patient Ages")
plt.xlabel("Age")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()

# Gender count plot
plt.figure(figsize=(7, 5))
sns.countplot(data=clean_df, x="gender")
plt.title("Patient Records by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
# Age distribution by gender
plt.figure(figsize=(8, 5))
sns.boxplot(data=clean_df, x="gender", y="age")
plt.title("Age Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Age")
plt.tight_layout()
plt.show()

# Gender versus diagnosis category, if available
if "Diagnosis Category" in clean_df.columns:
    gender_diagnosis = pd.crosstab(
        clean_df["Diagnosis Category"],
        clean_df["gender"]
    )

    display(gender_diagnosis)

    gender_diagnosis.plot(
        kind="bar", figsize=(12, 6)
    )
    plt.title("Diagnosis Category by Gender")
    plt.xlabel("Diagnosis Category")
    plt.ylabel("Number of Records")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
# Groupby analysis
group_summary = clean_df.groupby("gender", dropna=False).agg(
    Number_of_Records=("patient_id", "count"),
    Average_Age=("age", "mean"),
    Median_Age=("age", "median")
)

print("Summary by gender:")
display(group_summary.round(2))

# Pivot table: age group by gender
pivot_table = pd.pivot_table(
    clean_df,
    index="age_band",
    columns="gender",
    values="patient_id",
    aggfunc="count",
    fill_value=0,
    observed=False
)

print("Pivot table:")
display(pivot_table)
```

```python
# Select numerical variables only
numeric_columns = [
    col for col in ["age", "age_group", "gender_numeric"]
    if col in clean_df.columns
]

numeric_data = clean_df[numeric_columns].apply(
    pd.to_numeric, errors="coerce"
)

correlation = numeric_data.corr()

print("Correlation matrix:")
display(correlation.round(2))

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Bar chart: top 10 diagnoses
top_diagnoses = clean_df["Diagnosis"].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_diagnoses.sort_values().plot(kind="barh")
plt.title("Top 10 Diagnoses in the Dataset")
plt.xlabel("Number of Records")
plt.ylabel("Diagnosis")
plt.tight_layout()
plt.show()

# Scatter plot: age versus symptom count
if "symptom_count" in clean_df.columns:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        data=clean_df,
        x="age",
        y="symptom_count",
        hue="gender"
    )
    plt.title("Age versus Number of Listed Symptoms")
    plt.xlabel("Age")
    plt.ylabel("Number of Listed Symptoms")
    plt.tight_layout()
    plt.show()

# Pie chart: gender distribution
gender_counts = clean_df["gender"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Gender Distribution")
plt.tight_layout()
plt.show()
```
